# Run chatbot.py

This notebook runs the existing `chatbot.py` script programmatically and captures its output using Python.

```python
import subprocess
import sys
from pathlib import Path

script_path = Path(r"c:\Users\ricar\OneDrive\Documentos\ChatbotGemini\chatbot.py")
result = subprocess.run([sys.executable, str(script_path)], input="0\n", capture_output=True, text=True)
print('returncode:', result.returncode)
print('stdout:')
print(result.stdout)
print('stderr:')
print(result.stderr)
```
